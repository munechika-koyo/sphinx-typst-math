# Typst math in an nbsphinx notebook

This notebook is part of the Sphinx toctree. Its Markdown cells use the same native Typst syntax and the same `conf.py` package configuration as the `.md` pages.

## Native Typst

Inline math works inside a sentence: $sum_(k=1)^n k = (n(n+1))/2$.

A display equation can use Typst's matrix syntax:

$$
A = mat(1, 2; 3, 4)
$$

## Packages configured by Sphinx

The `physica` import from `conf.py` is also visible here:

$$
pdv(f, x, y, [1, 2])
$$

The project-wide `quick-maths` shorthands apply as well:

$$
x^2 = 9 quad <==> quad x = +-3
$$

## LaTeX migration with MiTeX

MiTeX can render an existing LaTeX fragment while the surrounding equation remains Typst source:

$$
#mi(raw("\\frac{1}{2}\\sum_{k=1}^{n} k"))
$$

> **Preview note:** Jupyter's live Markdown renderer may treat dollar math as LaTeX. Build the Sphinx documentation to preview these cells with the Typst renderer.
